 **<font color=darkgoldenrod>應用問題與背景理論</font>**：
 在保險精算中，理賠金額往往不是單一分配能很好描述的，譬如：理賠金額可能是下列兩種不同的型態：

1. 小額理賠：例如車險中常見的輕微刮傷，金額通常較低，分布集中。
1. 大額理賠：例如重大事故或醫療理賠，金額可能非常高，分布尾部很重。
如果只用一個伽瑪分配，可能無法同時捕捉「小額集中」與「大額重尾」的特性。這時候就可以用混合伽瑪分配：
$$
f(x)=\pi _1\cdot \mathrm{Gamma}(x;\alpha _1,\beta _1)+\pi _2\cdot \mathrm{Gamma}(x;\alpha _2,\beta _2)$$
其中：
- $\pi _1,\pi _2$：混合權重 (例如小額理賠佔 80%，大額理賠佔 20%)
- $\alpha _1,\beta _1$：小額理賠的伽瑪分配參數
- $\alpha _2,\beta _2$：大額理賠的伽瑪分配參數

這樣的模型能同時描述兩種不同型態的理賠金額分配。本文採 MLE 方法估計混合伽瑪分配的參數，並透過模擬實驗評估其在不同情境下的表現。其中對數概似函數為：
$$\ell (\Omega)=\sum _{i=1}^n \ln \left[\pi _1\cdot \mathrm{Gamma}(x_i;\alpha _1,\beta _1)+\pi _2\cdot \mathrm{Gamma}(x_i;\alpha _2,\beta _2)\right]$$

其中 $$\Omega=\{\pi _1, \pi _2, \alpha _1,\beta _1,\alpha _2,\beta _2 | \pi_1 + \pi_2 = 1, 0 \leq \pi_1 \leq 1, \alpha_1, \beta_1, \alpha_2, \beta_2 > 0 \}$$。實驗計畫的安排如下：

- 自行設定資料生成的參數 $\Omega=\{\pi_1,\alpha_1,\beta_1, \alpha_2,\beta_2\}$，譬如兩個情況：
    - 一組之兩個分配較接近（視覺上好像只有一組）。
    - 另一組分開較遠些（視覺上看出兩個分配之混合）。
- 總樣本數 $n= 100, 300, 500, 1000, 2000$ 藉以評估樣本數大小對估計值的影響。
- 進行試做，以確定所選的組合參數適當，且程式碼能正確執行。這個階段可以圖、表的視覺化方式來呈現。
- Monte Carlo 模擬的實驗次數為 $N=10000$（估計執行時間 3 ~ 4 小時，不過演算法或初始值的選擇會影響實際時間）。
- 建立三個 pandas 表格以評估 5 個參數的估計結果，分別是：
    - Mean：N 次實驗的估計值的平均數，$Mean(\hat{\theta})=\frac{1}{N}\sum_{k=1}^{N} \hat{\theta}_k=\bar{\theta}$
    - Bias：N 次實驗的估計值的平均數減去真實值，$Bias(\hat{\theta})=\bar{\theta}-\theta$
    - MSE：N 次實驗的估計值的均方誤差（Mean Square Error, MSE），$MSE(\hat{\theta})=\frac{1}{N}\sum_{k=1}^{N} (\hat{\theta}_k - \theta)^2$。
- 從上述的數據呈現 MLE 估計的一致性特色（consistency）：隨著樣本數 $n$ 的增加，估計值逐漸收斂於真實值。
- （optional）MLE 估計的漸進有效性（asymptotic efficiency）：在樣本數 $n$ 很大時，MLE 估計值的變異數達到 Cramér-Rao 下界，表示 MLE 是最有效率的估計方法。比較粗淺的做法是觀察估計值的變異數隨著樣本數的增加而減小。
- （optional）將上述實驗做成 GUI app，讓使用者可以自行設定參數組合、樣本數、實驗次數，並呈現估計結果的圖表與表格。

注意：為避免 Monte Carlo 得到的估計值存在 label switching 的問題，建議在每次估計後，檢查排序，以確保兩個分配的參數順序一致。排序的法則可以採

$$\frac{\alpha_1}{\alpha_1+\beta_1} < \frac{\alpha_2}{\alpha_2+\beta_2}.$$

#### <font color=skyblue>驗證 MLE 估計的特性</font>（供參考）

<font color="yellow">Consistency of MLEs</font>
The consistency of MLEs is a property that ensures that as the sample size increases, the maximum likelihood estimator converges in probability to the true parameter value. This means that with a sufficiently large sample size, the MLE will provide an accurate estimate of the underlying parameter.


用 Monte Carlo 驗證

$$P\bigl( \|\hat{\theta}_n-\theta_0\|>\varepsilon \bigr) \to 0$$

這比只看 bias 或 variance 更接近 consistency 的數學定義。對於多維參數 ，可以直接畫

$$\|\hat{\theta}_n-\theta_0\|​$$

的平均值或分位數隨 $n$ 的變化，這是統計模擬研究中很常見的做法。

<font color=yellow>Asymptotic efficiency of MLEs</font>
The asymptotic efficiency of MLEs is a measure of how well the maximum likelihood estimator performs as the sample size increases. It is often compared to the Cramér-Rao lower bound, which provides a theoretical limit on the variance of any unbiased estimator.


$$\sqrt{n}(\hat{\theta}_n-\theta_0) \overset{d}{\to} N(0, I^{-1}(\theta_0))$$

其中 $I(\theta_0)$ 是 Fisher information matrix。換句話說，MLE 不只是收斂而已，它的大樣本變異數還達到了 Cramér-Rao Lower Bound

$$Var(\hat{\theta}_{MLE}) \approx \frac{1}{n} I(\theta_0)^{-1}$$

用 Monte Carlo 驗證

估計 Monte Carlo covariance

1. 產生 $N$ 個 MLE 估計值 $\hat\theta^{(1)}, \dots, \hat\theta^{(N)}$。
2. 計算這些估計值的 sample covariance 來近似 MLE 的 asymptotic covariance：

$$\widehat\Sigma_{MC} = \mathrm{Cov} \left( \hat\theta^{(1)}, \dots, \hat\theta^{(N)} \right).$$
3. 計算 Fisher information matrix $I(\theta_0)$，並計算其逆矩陣 $I(\theta_0)^{-1}$。

計算

$$H(\hat\theta) = - \frac{\partial^2 \ell(\theta)} {\partial\theta\partial\theta^T} \Biggr|_{\hat\theta}.$$

則

$$\widehat I = \frac1M \sum_{i=1}^{N} H(\hat\theta^{(i)}).$$

4. 比較 $n\widehat\Sigma_{MC}$ 與 $I^{-1}$，檢查它們是否接近。這可以通過計算兩個矩陣的 Frobenius 範數差異或其他矩陣距離度量來完成。
