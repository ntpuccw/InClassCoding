 **<font color=darkgoldenrod>背景理論與問題</font>**：
 統計應用上經常需要檢定某筆資料是否來自常態分配，假說如下：

$$
H_0: data \sim N(\mu, \sigma^2)\\
H_a: data \not\sim N(\mu, \sigma^2)
$$

Jarque-Bera (JB) 檢定是一個常用的常態檢定方法，其檢定統計量根據樣本的偏態係數（skewness）與峰度（kurtosis）來衡量資料偏離常態分配的程度。JB 統計量的計算公式如下：

$$
JB = \frac{n}{6} \left( S^2 + \frac{(K-3)^2}{4} \right)
$$

其中 $S$ 為樣本的偏態係數，$K$ 為樣本的峰度，$n$ 為樣本大小。 在大樣本下，JB 統計量服從自由度為 2 的卡方分配。為觀察 JB 統計量的表現，我們將進行以下幾個實驗：

1. 首先，將 JB 統計量拆成兩部分：偏態部分和峰度部分，即 $$JB = JB_{skewness} + JB_{kurtosis}$$ 其中 $$JB_{skewness} = \frac{n}{6} S^2$$, $$JB_{kurtosis} = \frac{n}{24} (K-3)^2$$
透過蒙地卡羅模擬，分別觀察這兩個部分在不同樣本數下的分配情形，並比較其與理論分配（卡方分配自由度 1）的吻合度。

1. 接著，觀察整體 JB 統計量在不同樣本數下的分配情形，並與理論的卡方分配進行比較。
1. 為了檢驗 JB 檢定的顯著水準維持（the maintenance of the significance level），設定在不同樣本數下，模擬資料來自常態分配（即服從 $H_0$），並計算顯著水準 $\alpha = 0.05$ 下，JB 檢定拒絕虛無假設的比例。
1. 最後，為了評估 JB 檢定在面對不同母體分配時的檢定力（power），我們將模擬資料來自非常態分配，並計算在不同樣本數和顯著水準 $\alpha = 0.05$，JB 檢定拒絕虛無假設的比例。此外，我們也將與其他常見的常態檢定方法（如 Kolmogorov-Smirnov 檢定、Shapiro-Wilk 檢定、、、）進行比較，以評估 JB 檢定統計量的相對表現。
