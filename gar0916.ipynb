<font color=yellow> Numerical Integration</font>

```python
# compute the integral of f(x) = x^2 + 3x + 5 from 1 to 8

import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt

a = -2
b = 6
# f(x) = sqrt(x^2 + 4x + 12)
# f = lambda x: np.sqrt(x**2 + 4*x + 12)
def f(x):
    return np.sqrt(x**2 + 4*x + 12)

# draw the function and the area under the curve
plt.figure(figsize=(6, 3))
x_plot = np.linspace(a-2, b+2, 400)
y_plot = f(x_plot)

x_fill = np.linspace(a, b, 200)
y_fill = f(x_fill)

plt.plot(x_plot, y_plot, label=r"$f(x) = \sqrt{x^2 + 4x + 12}$", color="blue")
plt.fill_between(x_fill, y_fill, color="blue", alpha=0.25)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Numerical Integration of $f(x) = \sqrt{x^2 + 4x + 12}$")
plt.legend()
plt.grid(True)
plt.show()

result = integrate.quad(f, a, b)

print(f"Numerical integral of f(x) = sqrt(x**2 + 4*x + 12) from {a} to {b}: {result[0]:.4f}")
# print(f"Estimated absolute error: {error:.4f}")
```

<font color=yellow> Symbolic Integration</font>

```python
# compute the integral of f(x) = x^2 + 3x + 5 from 1 to 8 in symbolic form

import sympy as sp

x = sp.Symbol('x')
# f(x) = sp.sqrt(x**2 + 4*x + 12)
f = sp.sqrt(x**2 + 4*x + 12)
# f = x**2 + 3*x + 5

# compute the indefinite integral
indefinite_integral = sp.integrate(f, x)
print(f"Indefinite integral: {indefinite_integral}")

result = sp.integrate(f, (x, 1, 8))
print(f"Symbolic integral of f(x) = x**2 + 3*x + 5 from 1 to 8: {result}")

```

```python
# compute the definite integral of the standard normal distribution 
# function from -5 to 5

import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt

def f(x):
    return (1/np.sqrt(2*np.pi)) * np.exp(-x**2 / 2)

# draw the function and the area under the curve
plt.figure(figsize=(6, 3))
x_plot = np.linspace(-5, 5, 400)
y_plot = f(x_plot)

a=-1.96
b=1.96
x_fill = np.linspace(a, b, 200)
y_fill = f(x_fill)

plt.plot(x_plot, y_plot, label=r"$f(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}$", color="blue")
plt.fill_between(x_fill, y_fill, color="blue", alpha=0.25)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Numerical Integration of $f(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}$")
plt.legend()
plt.grid(True)
plt.show()

result = integrate.quad(f, a, b)

print(f"Numerical integral of f(x): {result[0]:.4f}")
```

```python
# Draw the function F(x) = integral of f(x) from -infinity to x

import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt

# f(x) is the standard normal distribution function
def f(x):
    return (1/np.sqrt(2*np.pi)) * np.exp(-x**2 / 2)

def F(x):
    return integrate.quad(f, -np.inf, x)[0]


# draw the function and the area under the curve
plt.figure(figsize=(6, 3))
x_plot = np.linspace(-5, 5, 400)

y_plot = np.array([F(x) for x in x_plot])

plt.plot(x_plot, y_plot, label=r"$f(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}$", color="blue")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Function $f(x) = \frac{1}{\sqrt{2\pi}} e^{-x^2/2}$")
plt.legend()
plt.grid(True)
plt.show()
```

<font color=yellow>Numerical Differentiation</font>

```python
# draw the function f(x) = ln(x)/x^3 from 1 to 10

import numpy as np
import matplotlib.pyplot as plt
# from scipy.misc import derivative

x = np.linspace(0.95, 6, 1000)

def f(x):
    return np.log(x) / x**3

y = f(x)

# use package scipy to compute f'(x) in numerical form
# def f_prime(x):
#     from scipy.misc import derivative
#     return derivative(f, x, dx=1e-6)

def f_prime(x):
    h = 1e-5
    return (f(x + h) - f(x - h)) / (2 * h)

y_prime = f_prime(x)

fig = plt.figure(figsize=(6, 3))
plt.plot(x, y, label="f(x)")
plt.plot(x, y_prime, label="f'(x)")
plt.xlabel("x")
plt.ylabel("f(x) and f'(x)")
plt.title("Function f(x) = ln(x)/x^3 and its derivative")
plt.legend()
plt.grid(True)
plt.show()
```

<font color=yellow> Solve </font>$f(x) = 0$

```python
# solve f(x)=0

import numpy as np
from scipy import optimize
import matplotlib.pyplot as plt

def f(x):
    return x - np.exp(-x/2)

x0 = 10
result = optimize.fsolve(f, x0)
print(f"Solutions to f(x)=0: {result}")

# draw the function f(x) = x - exp(-x/2) and its roots

x = np.linspace(0, 5, 1000)
y = f(x)

fig = plt.figure(figsize=(6, 3))
plt.plot(x, y, label="f(x)")
# mark the roots on the graph
for root in result:
    plt.plot(root, f(root), 'rx', label=f"Root: {root:.4f}")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Function f(x) = x - exp(-x/2) and its roots")
plt.legend()
plt.grid(True)
plt.show()
```

<font color=yellow>Solve</font> $\min_x f(x)$

- Let $f(x) = x^4 - 8x^3+16x^2-2x+8$. Solve $$\min_x f(x)$$
- $\displaystyle \min_{x \in (0,3)} \tan^{-1}\frac{5}{x}+\tan^{-1}\frac{2}{3-x}$

```python
# find the minimum of f(x) = atan(5/x) + atan(2/(3-x))

import numpy as np
from scipy import optimize
import matplotlib.pyplot as plt

def f(x):
    return np.arctan(5/x) + np.arctan(2/(3-x))

result = optimize.minimize_scalar(f, bounds=(0, 3), method="bounded")
x_min = result.x

print(f"Minimum of f(x) = atan(5/x) + atan(2/(3-x)): {x_min:.4f}")
print(f"Minimum value: {f(x_min):.4f}")

# draw the function f(x) = atan(5/x) + atan(2/(3-x)) and its minimum

x = np.linspace(0.01, 2.99, 1000)
y = f(x)

fig = plt.figure(figsize=(6, 3))
plt.plot(x, y, label="f(x)")
# mark the minimum on the graph
plt.plot(result.x, f(result.x), 'rx', label=f"Minimum: {result.x:.4f}")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Function f(x) = atan(5/x) + atan(2/(3-x)) and its minimum")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes

xmin, xmax = -1, 5
x = np.linspace(xmin, xmax, 100)
p = [1, -8, 16, -2, 8]
f = lambda x: np.polyval(p, x)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(x, f(x), linewidth=2, color="g")
# attach an inset axis object to the figure and use it to plot the function in the range of [3.7, 4.2]
axins = inset_axes(ax, width="30%", height="30%", loc="upper right")
x1 = np.linspace(3.8, 4.2, 100)
axins.plot(x1, f(x1), linewidth=1, color="g")
# draw a horizontal line at y=0
axins.plot([3.7, 4.2], [0, 0], linewidth=1, color="r")

ax.grid(True)
ax.set_xlabel("X"), ax.set_ylabel("f(x)")
ax.set_title("$f(x) = x^4 - 8x^3 + 16x^2 - 2x + 8$")
plt.show()
```
