### <font color=skyblue>檢定統計量的機率分布</font>

檢定統計量的機率分布是指在假設檢定中，當虛無假設成立時，檢定統計量所遵循的機率分布。這個分布可以用來計算檢定統計量的臨界值和 p 值，從而決定是否拒絕虛無假設。不同的檢定方法會對應不同的檢定統計量及其分布，例如 t 檢定使用 t 分布，卡方檢定使用卡方分布等。
<hr>

0. <font color=yellow>雙樣本 t 檢定 (Two-sample t-test)：</font>

$$H_0: \mu_1 = \mu_2$$
$$H_a: \mu_1 \neq \mu_2$$

檢定統計量的公式如下：
$$   
t = \displaystyle\frac{\bar{x}_1 - \bar{x}_2}{\sqrt{s_1^2/n_1 + s_2^2/n_2}} $$

其中，
- $\bar{x}_1$ 和 $\bar{x}_2$ 分別是兩組樣本的平均值
- $s_1^2$ 和 $s_2^2$ 分別是兩組樣本的變異數
- $n_1$ 和 $n_2$ 分別是兩組樣本的大小

當虛無假設成立時，$t$ 統計量遵循自由度為 $n_1 + n_2 - 2$ 的 $t$ 分布。

<font color=yellow>實驗 1: 統計量的抽樣分配與理論分佈的比較</font>

在這個實驗中，我們將模擬雙樣本 $t$ 檢定的統計量分佈，並觀察其與理論 $t$ 分配的吻合程度。首先假設兩組樣本來自相同的常態分配，譬如  $\mathcal{N}(0, 1)$, 即服從虛無假設 $H_0: \mu_1 = \mu_2$。

想實驗觀察 $t$ 統計量的分佈與樣本數的關係及與理論分佈的吻合程度。
- 樣本數 $n_1 = n_2 = 10, 20, 30, 50, 100, 500$。
- 重複抽樣 $N=10000$ 次，計算每次抽樣的 $t$ 統計量，並繪製其分佈圖。
- 嘗試使用公式計算統計量與使用 `scipy.stats.ttest_ind` 計算統計量，觀察兩者是否一致。

```python
# compute t statistic for two independent samples from normal distributions with equal variance

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import ttest_ind, t

# take n samples from a normal distribution with mean 0 and standard deviation 1
# repeat this sampling N times to create two independent samples
N = 10000
n = 40
samples1 = np.random.normal(0, 1, size=(n, N))
samples2 = np.random.normal(0, 1, size=(n, N))

# compute t statistic
t_statistic, p_value = ttest_ind(samples1, samples2)

# plot histogram of t statistic
fig = plt.figure(figsize=(6, 3))
plt.hist(t_statistic, bins=100, density=True, alpha=0.5, edgecolor='black')
# plot the theoretical t distribution with df = 2n - 2
df = 2 * n - 2
xx = np.linspace(-4, 4, 100)
plt.plot(xx, t.pdf(xx, df), 'r', linewidth=2, label='t distribution (df={})'.format(df))
plt.xlabel('t statistic')
plt.ylabel('Density')
plt.title('Histogram of t statistic')
# add a legend
plt.legend(['t statistic', 't distribution'])
# fix the x-axis limits
plt.xlim(-4, 4)
plt.show()
# print t statistic and p value
# print(t_statistic, p_value)
```

<font color=yellow>實驗 3: 型一誤與檢定力</font>


- 型一誤 (Type I Error) $\alpha$ 為在虛無假設為真時，錯誤地拒絕虛無假設的機率。型一誤率通常表示為

$$\alpha = P_{rob}(\text{拒絕 } H_0 | H_0 \text{ 為真})$$

型一誤又稱為顯著水準 (significance level)，通常設定為 0.05 或 0.01，做為判斷檢定結果是否顯著的標準。

- 檢定力 (Power) 為在對立假設為真時，成功地拒絕虛無假設的機率，即

$$Power = P_{rob}(\text{拒絕 } H_0 | H_a \text{ 為真})$$

檢定力越高，表示檢定方法越能有效地識別出虛無假設為假的情況，也稱為敏感度 (sensitivity)。檢定力與樣本數、顯著性水準等因素有關。

```python
# compute the power of the t-test for N = 100000 and n = 40

N = 100000
n = 10
alf = 0.05
samples1 = np.random.normal(0, 1, size=(n, N))
samples2 = np.random.normal(0, 1, size=(n, N))

# compute t statistic
t_statistic, p_value = ttest_ind(samples1, samples2)

power = np.mean(p_value < alf)
print('Power of the t-test for N = {} and n = {}: {:.4f}'.format(N, n, power))
```

broadcasting vs looping

```python
# Broadcasting vs Looping
# take n samples from a normal distribution with mean 0 and standard deviation 1
# Do this N times by looping
# Do this N times by broadcasting

import time
import numpy as np
import matplotlib.pyplot as plt

N = 100000
n = 100

# by broadcasting
# time this code and compare it to the loop version below
start = time.time()
samples = np.random.normal(0, 1, size=(N, n))
mean = np.mean(samples, axis=1)
std = np.std(samples, axis=1)
end = time.time()
print('By broadcasting:{:.4f}s'.format(end - start))

# by looping and storing the results in a numpy array
start = time.time()
means = np.zeros(N)
stds = np.zeros(N)
for i in range(N):
    samples = np.random.normal(0, 1, size=n)
    mean = np.mean(samples)
    std = np.std(samples)
    means[i] = mean
    stds[i] = std

end = time.time()
print('By looping:{:.4f}s'.format(end - start))
```
