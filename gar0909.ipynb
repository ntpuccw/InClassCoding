### Numerical Integration

This notebook demonstrates numerical integration using `scipy.integrate.quad`:

1. **Compute the integral** of $f(x) = x^2 + 3x + 5$ over $[a, b]$, printing the result and estimated error rounded to 4 decimal places.
2. **Visualize the function** by plotting $f(x)$ over a padded range, shading the area under the curve between $a$ and $b$ to represent the integral, and displaying the computed integral value at the center of the shaded region.

```python
import numpy as np
from scipy import integrate

def f(x):
    return x**2 + 3*x + 5

a, b = 1, 8

result, error = integrate.quad(f, a, b)
print(f"Numerical integral of f(x) = x**2 + 3*x + 5 from {a} to {b}: {result:.4f}")
print(f"Estimated absolute error: {error:.4f}")
```

```python
import matplotlib.pyplot as plt

x_plot = np.linspace(a - 1, b + 1, 400)
y_plot = f(x_plot)

x_fill = np.linspace(a, b, 200)
y_fill = f(x_fill)

plt.plot(x_plot, y_plot, label=r"$f(x) = x^2 + 3x + 5$", color="blue")
plt.fill_between(x_fill, y_fill, color="skyblue", alpha=0.5, label=f"Integral area [{a}, {b}]")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)

x_mid = (a + b) / 2
y_mid = f(x_mid) / 2
plt.text(x_mid, y_mid, f"{result:.4f}", ha="center", va="center", fontsize=12, fontweight="bold")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Numerical Integration of $f(x) = x^2 + 3x + 5$")
plt.legend()
plt.grid(True)
plt.show()
```
