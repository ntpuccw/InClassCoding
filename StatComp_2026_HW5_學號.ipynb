 **<font color=darkgoldenrod>背景理論與實驗項目</font>**：

在可靠度工程中，元件可靠度分析是保障系統穩定運作的關鍵。元件通常採用串聯或並聯方式組合，這兩種拓撲結構對系統整體可靠度產生顯著影響。串聯組合中任一元件失效即導致整個系統失效，而並聯組合則透過冗餘設計提升系統可靠度。本實驗分析串聯與並聯系統的可靠度特性如下：

1. 兩個不同元件串聯後的可靠度分析與近似貝他分配參數估計：
    - 令 $Z=XY$ 代表兩個不同元件 $X, Y$ 串聯後之複合系統的可靠度隨機變數。
    - 從數理統計角度推導轉換後隨機變數的概率密度函數 $f(z)$。
    - 採用最小平方誤差法 (LSE) 以貝他分配逼近 $f(z)$： 建立 PDF 擬合最小值問題：$$\displaystyle \min_{a,b>0} \int_0^1 (f(z) - \beta(z|a,b))^2 \;dz$$

    - 基於 $Z$ 的抽樣分配，分別以 MLE 與 MM 法估計貝他分配參數 $(a, b)$。
    
2. 兩個相同元件並聯系統的可靠度分析與貝他分配參數估計：
    - 令  $Z=1-(1-X)^2$ 代表兩個相同元件 $X$ 並聯後之複合系統的可靠度隨機變數。
    - 從數理統計角度推導轉換後隨機變數的概率密度函數 $f(z)$。
    - 採用最小平方誤差法 (LSE) 以貝他分配逼近 $f(z)$： 建立 PDF 擬合最小值問題：$$\displaystyle \min_{a,b>0} \int_0^1 (f(z) - \beta(z|a,b))^2 \;dz$$
    - 基於 $Z$ 的抽樣分配，分別以 MLE 與 MM 法估計貝他分配參數 $(a, b)$。
3. 混合系統的可靠度分析：自行設計涵蓋串聯與並聯元件的複合系統，估計整個混合系統的貝他分配參數。
