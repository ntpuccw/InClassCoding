本專題使用最大概似估計法（Maximum Likelihood Estimation, MLE）進行常態分配函數的參數估計實驗。實驗依未知參數的數量分兩個部分：一、僅含單一參數 $\mu$，二、含 $\mu$ 及 $\sigma$ 兩個參數。分別繪製並觀察聯合概似函數圖的長相，並進一步計算聯合概似函數之最大值。同時，也觀察樣本數對繪製聯合概似函數圖及計算其最大值的影響。


<hr>

 **<font color=darkgoldenrod>背景理論與問題</font>**：

令 $x_1, x_2, \cdots, x_n$ 為 $n$ 個來自常態分配 $N(\mu, \sigma^2)$ 的獨立樣本。本文依下列情境執行參數估計實驗：

1. 假設 $\sigma$ 已知（譬如 $\sigma=1$）， $\mu$ 未知。其聯合概似函數為

    $$L(\mu|x_1, x_2, \cdots, x_n)=\Pi_{i=1}^{n} f(x_i; \mu, \sigma=1)$$

    其中 $f(x; \mu, \sigma=1)$ 代表常態分配 $N(\mu, \sigma=1)$ 的機率密度函數（PDF）。

1. 假設 $\mu, \sigma$ 皆未知，其聯合概似函數為

    $$L(\mu, \sigma|x_1, x_2, \cdots, x_n)=\Pi_{i=1}^{n} f(x_i; \mu, \sigma)$$

    其中 $f(x; \mu, \sigma)$ 代表常態分配 $N(\mu, \sigma)$ 的機率密度函數（PDF）。
