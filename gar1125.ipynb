#### <font color="skyblue"> Exercise: Constraint optimization problem with integrations in the objective function</font>  #### 

A Least Squared Error (LSE) problem

$$ \min_{a,b>0} \int_0^1 (f(z) - \beta(z|a,b))^2 \;dz$$

where

- $Z = 1 - (1-X)^2$
- $f(z)=\frac{f_X(1-\sqrt{1-z})}{2\sqrt{1-z}}$
- $X \sim \beta(a_1,b_1)$ 

Test the following cases: 
- $(a_1, b_1) = (18, 2)$
- $(a_1, b_1)=(17, 3)$ 

Note: $a, b$ can also be estiamted using the method of moments and maximum likelihood estimation by sampling from the distribution of $Z$.
