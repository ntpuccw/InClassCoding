#### <font color=skyblue>雙變量函數的繪圖 & 多變量函數的參數估計與極值計算</font>

<font color=yellow>雙變量函數的繪圖與最小值計算（1）</font>

- 繪製函數 $f(x,y) = xe^{-x^2-y^2}$ 的立體網格圖（mesh(wireframe), surface）與等高線圖（contour）。

- 計算函數 $f(x,y)$ 的最小值，即
$$\min_{x,y} f(x,y) = \min_{x,y} xe^{-x^2-y^2}$$

<font color=yellow>雙變量函數的繪圖與最小值計算（2）</font>

Let $$f(x_1,x_2) = (x_1-2)^4 + (x_1-2)^2x_2^2 + (x_2+1)^2$$.

- 繪製立體圖（wireframe）與等高線圖（contour）。

- 計算 $\displaystyle\min_{x_1, x_2} f(x_1, x_2)$.

<font color=yellow>雙變量函數的繪圖與最小值計算（3）</font>

著名的 Rosenbrock’s banana 測試函數

$$f(x, y) = 100(y-x^2)^2 + (1-x)^2$$

- 繪製立體圖（wireframe）與等高線圖（contour）。
- 計算 $\displaystyle\min_{x, y} f(x, y)$.

<font color=yellow>限制式條件下的最小值問題 The constraint minimization</font>

求解下列限制式條件的最小值

$$\displaystyle \min_{x_1, x_2 \in \Omega} (x_1-2)^4 + (x_1-2)^2x_2^2 + (x_2+1)^2$$

其中 $\Omega$ 分別為：

1. $\Omega = \{x_1, x_2 \in R|\; x_1 \geq 0, x_2 \geq 0 \}$

2. $\Omega = \{x_1, x_2 \in R|\; x_1 \leq 1, x_2 \leq 2 \}$

3. $\Omega = \{x_1, x_2 \in R|\; 0 \leq x_1 \leq 1, 0 \leq x_2 \leq 2 \}$

4. $\Omega = \{x_1, x_2 \in R|\; 0 \leq x_1 \leq \infty, -\infty \leq x_2 \leq 2 \}$

5. $\Omega = \{x_1, x_2 \in R|\; x_1 +x_2 \leq 0.9 \}$

6. $\Omega = \{x_1, x_2 \in R|\; 1.5 \leq x_1 +x_2 \leq 2 \}$

7. $\Omega = \{x_1, x_2 \in R|\;  \sqrt{x_1^2 +x_2^2} \leq 1 \}$

8. $\Omega = \{x_1, x_2 \in R|\;  \sqrt{x_1^2 +x_2^2} \leq 1, \; x_1x_2 \geq 0\}$
