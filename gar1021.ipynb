<font color=yellow>雙變量函數的繪圖與最小值計算（1）</font>

- 繪製函數 $f(x,y) = xe^{-x^2-y^2}$ 的立體網格圖（mesh(wireframe), surface）與等高線圖（contour）。

- 計算函數 $f(x,y)$ 的最小值，即
$$\min_{x,y} f(x,y) = \min_{x,y} xe^{-x^2-y^2}$$

```python
# draw the wireframe for f(x,y) = x*exp(-x^2 - y^2)

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# Make data.
x = np.arange(-3, 3, 0.1)
y = np.arange(-3, 3, 0.1)
X, Y = np.meshgrid(x, y)
Z = X * np.exp(-X**2 - Y**2)

# Plot the wireframe.
ax.plot_wireframe(X, Y, Z, cmap=cm.coolwarm,
    rstride=1, cstride=1, linewidth=0.5)

ax.view_init(10, -60)  #(elev=-165, azim=60)
# Add a color bar which maps values to colors.
# fig.colorbar(surf, shrink=0.5, aspect=5)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
plt.show()
```

```python
# plot the contour of the function f(x,y) = x*exp(-x^2 - y^2)

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(111)
# Make data.
x = np.arange(-2, 2, 0.05)
y = np.arange(-2, 2, 0.05)
X, Y = np.meshgrid(x, y)
Z = X * np.exp(-X**2 - Y**2)

# Plot the contour with a color map and 10 contour levels.
# ax.contour(X, Y, Z, cmap=cm.coolwarm, levels=10)
# add color bar
plt.colorbar(ax.contour(X, Y, Z, cmap=cm.coolwarm, levels=100))
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
plt.show()
```

```python
# compute the minimum of the function f(x,y) = x*exp(-x^2 - y^2) using scipy.optimize.minimize_scalar

import numpy as np
import scipy.optimize as opt

def f(x):
    return x[0] * np.exp(-x[0]**2 - x[1]**2)

result = opt.fmin(func=f, x0=[0,-1])
# print the result of the optimization
print(f"Minimum of f(x,y) = x*exp(-x^2 - y^2): {result}")
print(f"Minimum value: {f(result)}")
```

<font color=yellow>雙變量函數的繪圖與最小值計算（2）</font>

Let $$f(x_1,x_2) = (x_1-2)^4 + (x_1-2)^2x_2^2 + (x_2+1)^2$$.

- 繪製立體圖（wireframe）與等高線圖（contour）。

- 計算 $\displaystyle\min_{x_1, x_2} f(x_1, x_2)$.

```python
# plot the contour of the function f(x,y) = (x-2)^4 + (x-2)^2*y^2 + (y+1)^2

import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

f = lambda x: (x[0] - 2)**4 + (x[0]-2)**2 * x[1]**2 + (x[1]+1)**2

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
# Make data.
x = np.arange(0, 4, 0.05)
y = np.arange(-3, 1, 0.05)
X, Y = np.meshgrid(x, y)
Z = f([X, Y])

ax.plot_wireframe(X, Y, Z, cmap=cm.coolwarm,
    rstride=1, cstride=1, linewidth=0.5)

ax.view_init(10, -70)  #(elev=-165, azim=60)
# Add a color bar which maps values to colors.
# fig.colorbar(surf, shrink=0.5, aspect=5)
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
plt.show()

# plot the contour of the function f(x,y) = (x-2)^4 + (x-2)^2*y^2 + (y+1)^2
fig = plt.figure(figsize=(6, 4))
ax = fig.add_subplot(111)
plt.colorbar(ax.contour(X, Y, Z, cmap=cm.coolwarm, levels=500))
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
plt.show()
```

<font color=yellow>雙變量函數的繪圖與最小值計算（3）</font>

著名的 Rosenbrock’s banana 測試函數

$$f(x, y) = 100(y-x^2)^2 + (1-x)^2$$

- 繪製立體圖（wireframe）與等高線圖（contour）。
- 計算 $\displaystyle\min_{x, y} f(x, y)$.

```python
# draw the wireframe for f(x,y) = 100*(y-x^2)^2 + (1-x)^2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import scipy.optimize as opt

f = lambda x: 100*(x[1]-x[0]**2)**2 + (1-x[0])**2

# compute the minimum of the function f(x,y) = 100*(y-x^2)^2 + (1-x)^2 
result = opt.fmin(func=f, x0=[0,0])
# print the result of the optimization
# print the result to decimal places 4
print(f"Minimum value: {f(result):.4f} at x={result[0]:.4f}, y={result[1]:.4f}")

x = np.linspace(-2, 2, 100)
y = np.linspace(-1, 3, 100)
X, Y = np.meshgrid(x, y)
Z = f([X, Y])

# plot a 1x2 grid of subplots, left subplot is the wireframe, right subplot is the contour
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
# ax1 nees to be a 3D subplot
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_wireframe(X, Y, Z)
ax1.set_xlabel('X axis')
ax1.set_ylabel('Y axis')
# ax1.set_zlabel('Z axis')

ax2.contour(X, Y, Z, cmap=cm.coolwarm, levels=100)
# add a cross at the minimum point
ax2.plot(result[0], result[1], 'ro', markersize=10)
ax2.set_xlabel('X axis')
ax2.set_ylabel('Y axis')
ax2.set_title('Contour Plot')

plt.show()
```

<font color=yellow>限制式條件下的最小值問題 The constraint minimization</font>

求解下列限制式條件的最小值

$$\displaystyle \min_{x_1, x_2 \in \Omega} (x_1-2)^4 + (x_1-2)^2x_2^2 + (x_2+1)^2$$

其中 $\Omega$ 分別為：

1. $\Omega = \{x_1, x_2 \in R|\; x_1 \geq 0, x_2 \geq 0 \}$

2. $\Omega = \{x_1, x_2 \in R|\; x_1 \leq 1, x_2 \leq 2 \}$

3. $\Omega = \{x_1, x_2 \in R|\; 0 \leq x_1 \leq 1, 0 \leq x_2 \leq 2 \}$

4. $\Omega = \{x_1, x_2 \in R|\; 0 \leq x_1 \leq \infty, -\infty \leq x_2 \leq 2 \}$

5. $\Omega = \{x_1, x_2 \in R|\; x_1 +x_2 \leq 0.9 \}$

6. $\Omega = \{x_1, x_2 \in R|\; 1.5 \leq x_1 +x_2 \leq 2 \}$

7. $\Omega = \{x_1, x_2 \in R|\;  \sqrt{x_1^2 +x_2^2} \leq 1 \}$

8. $\Omega = \{x_1, x_2 \in R|\;  \sqrt{x_1^2 +x_2^2} \leq 1, \; x_1x_2 \geq 0\}$

```python
# compute the constraint minimum of the function f(x,y) = (x-2)^4 + (x-2)^2*y^2 + (y+1)^2
# subject to the constraint x>=0, y>=0

import scipy.optimize as opt

f = lambda x: (x[0] - 2)**4 + (x[0]-2)**2 * x[1]**2 + (x[1]+1)**2

# Define the constraints
cons = [{'type': 'ineq', 'fun': lambda x: x[0]},  # x >= 0
        {'type': 'ineq', 'fun': lambda x: x[1]}]  # y >= 0

# Compute the constrained minimum
result = opt.minimize(f, x0=[0, 0], constraints=cons)

print(f"Constrained minimum value: {f(result.x):.4f} at x={result.x[0]:.4f}, y={result.x[1]:.4f}")
```

```python
# a new constraint optimization problem
# subject to x<=1, y<=2

cons = [{'type': 'ineq', 'fun': lambda x: 1 - x[0]},     # x <= 1
        {'type': 'ineq', 'fun': lambda x: 2 - x[1]}]  # y <= 2

result = opt.minimize(f, x0=[0, 0], constraints=cons)
print(f"Constrained minimum value: {f(result.x):.4f} at x={result.x[0]:.4f}, y={result.x[1]:.4f}")

```
