<font color=yellow>Define constants</font>

```python
N = 10000 # number of Monte Carlo samples to generate for each sample
n = [10, 20, 30, 50, 100, 300, 500] # sample sizes to test
```

```python
alfa = 0.05

# generate data from a given distribution
distribution = 'Normal'  # choose a distribution from the distributions
sample_data = distributions[distribution](size=(N, 30))

# JB test
result_JB = JB_statistic(sample_data)
# compute rejection rates
rejection_rates = np.mean(result_JB['p_value'] < alfa, axis=0)
print(f"Rejection rates for 'JB' test on {distribution} distribution: {rejection_rates}")

# Shapiro-Wilk test
result_Shapiro = normality_test(sample_data, 'shapiro', alpha=alfa)
rejection_rates = np.mean(result_Shapiro['p_value'] < alfa, axis=0)
print(f"Rejection rates for 'Shapiro' test on {distribution} distribution: {rejection_rates}")
```
